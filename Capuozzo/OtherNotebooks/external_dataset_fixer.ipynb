```python
import os
import shutil
import pandas as pd
```

```python
path = "../../RSNA-BTC-Datasets/UPENN-GBM"
```

```python
unmeth = os.listdir(path+"_mat/T1wCE/0")
unmeth_list = [x.split("-")[2].split("_")[0] for x in unmeth]
meth = os.listdir(path+"_mat/T1wCE/1")
meth_list = [x.split("-")[2].split("_")[0] for x in meth]
print("Unmethylated:")
print(unmeth_list)
print("Methylated:")
print(meth_list)
merged_list = unmeth_list + meth_list
merged_list.sort()
mgmt_values = []
for elem in merged_list:
    if elem in unmeth_list:
        mgmt_values.append(0)
    else:
        mgmt_values.append(1)
merged_df = pd.DataFrame({"BraTS21ID": merged_list, "MGMT_value": mgmt_values})
print(merged_df)
merged_df.to_csv(f"upenn_train_labels.csv", index=False)
```

```python
for folder in os.listdir(path):
    subfolders = os.listdir(path+"/"+folder)
    for subfolder in subfolders:
        for final_folder in os.listdir(path+"/"+folder+"/"+subfolder):
            source = path+"/"+folder+"/"+subfolder+"/"+final_folder
            dest = path+"/"+folder+"/"+subfolder[:-6]+"_"+final_folder
            shutil.move(source, dest)
```
